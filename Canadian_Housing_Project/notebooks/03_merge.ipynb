# Merge CMHC and CSI Data

I simply merge CMHC and CSI Data here. The EDA and visualizations are in notebook 4.

## Step 1: Import Libraries

```python
import pandas as pd
```

## Step 2: Load the Two Cleaned Files

```python
housing = pd.read_csv("../data/processed/cmhc_clean_2020_2024.csv")
csi = pd.read_csv("../data/processed/csi_clean_2020_2024.csv")

print("CMHC:", housing.shape, "\n CSI:", csi.shape)
```

## Step 3: Merge Them
Both files have a composite key in the `centre`/`year` column now. I ended up with the number of rows I expected (17 cities x 5 years = 85).

```python
df = housing.merge(csi, on=["centre", "year"], how="inner")

print("merged shape:", df.shape)

# make sure nothing got silently dropped by ensuring both CSV's have the exact same 17 centre name spellings
dropped_housing = set(housing["centre"]) - set(df["centre"])
dropped_csi = set(csi["centre"]) - set(df["centre"])
print("dropped from housing:", dropped_housing)
print("dropped from csi:", dropped_csi)

assert df.shape[0] == 85
assert df.isna().sum().sum() == 0

df.head()
```

## Step 4: Save the Merged File
So notebook 4 (and anything else) can just read this instead of re-merging.

```python
df.to_csv("../data/processed/housing_csi_merged_2020_2024.csv", index=False)
print("Saved!\nDF:", df.shape)
```

## Step 1: Load Data

```python
import pandas as pd

df = pd.read_csv("../data/processed/housing_csi_merged_2020_2024.csv")
per_centre_corr = pd.read_csv("../data/processed/per_centre_corr_2020_2024.csv")
```

## Step 2: National Trend
![national trend](../images/national_rent_vs_csi_trend.png)

```python
yearly = df.groupby("year")[["rent_rms", "csi_total"]].mean()

rent_change = yearly.loc[2024, "rent_rms"] / yearly.loc[2020, "rent_rms"] - 1
csi_change = yearly.loc[2024, "csi_total"] / yearly.loc[2020, "csi_total"] - 1

print(f"Avg. rent: ${yearly.loc[2020, 'rent_rms']:.0f} -> ${yearly.loc[2024, 'rent_rms']:.0f}  ({rent_change:+.1%})")
print(f"Avg. CSI:  {yearly.loc[2020, 'csi_total']:.1f} -> {yearly.loc[2024, 'csi_total']:.1f}  ({csi_change:+.1%})")
```

**What this shows:** There is no visible correlation between rent and crime severity by city. Québec CMA (CSI 55.5) and Ottawa-Gatineau's Ontario part (CSI 54.1) have nearly identical crime severity (16th and 17th), yet rent differs by 62% (＄1,159 vs. ＄1,880). Winnipeg has the highest crime severity (125.2) but ranks just 12th of 17 in rent (＄1,507, below both the mean and median). Inversely, Vancouver has the highest rent (＄2,314) but only 6th-highest crime severity (82.3); above average, but far from the lowest.

## Step 3: Cities Compared, 2024
![city comparison](../images/city_rent_vs_csi_latest_year.png)

```python
latest = df[df["year"] == df["year"].max()]
xsec_corr = latest["rent_rms"].corr(latest["csi_total"])
print(f"Cross-city correlation (rent vs. csi_total), 2024: {xsec_corr:.3f}")

latest.sort_values("rent_rms")[["centre", "rent_rms", "csi_total"]].reset_index(drop=True)
```

**What this shows:** There is no visible correlation between rent and crime severity by city. Quebec has similar low rates of crime to Ottawa-Gatineau and yet the rent difference is substantial. In Winnipeg, where the crime severity is highest, the rent is average among all 17 cities. Inversely, in Vancouver, where the rent is highest, the crime severity is average among all 17 cities.

## Step 4: Within-City Correlation
![within-city correlation](../images/within_city_correlation.png)

```python
per_centre_corr.sort_values("rent_csi_correlation")
```

**What this shows:** The spread of positive and negative correlation suggest the strength of these correlations (several above 0.8 in magnitude) shows rent and crime severity are often tightly linked within a given city, but the inconsistent direction across cities means there's no relationship that generalizes across the dataset.

## Step 5: Overall Takeaway
My findings using the current dataset are inconclusive at this time. There is some significant positive correlation (> 0.8) in 7 centres, but the remaining 10 show weaker or negative correlations. With such a small dataset (5 years as data points for 17 centres) and so few factors, rent and crime, there aren't other available statistics to consider at this time.

Other things to consider would be the impact of tourism, education, employment, income, and food security on rent and crime. Extending the analysis across a larger span of years would help as well, although data availability will likely be the limitation.
